```python
import numpy as np
import pandas as pd


data = {
    'Name': ['Anu', 'Ravi', 'Priya', 'Kiran', 'Meena', 'Arjun'],
    'Age': [21, 22, None, 23, 21, None],
    'Salary': [25000, None, 30000, 28000, None, 32000],
    'Department': ['CSE', 'ECE', 'CSE', None, 'ISE', 'ECE'],
    'City': ['Mysore', 'Bangalore', 'Mysore', 'Hubli', 'Bangalore', 'Mysore']
}
df= pd.DataFrame(data)
print(df)
```

```python
print(df.isnull())
```

```python
print(df.isnull().sum())
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
print(df)
```

```python
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
print(df)
```

```python
df['Department'] = df['Department'].fillna(df['Department'].mode()[0])
print(df)
```

```python
df["Age"]=df["Age"].ffill()
print(df)
```

```python
print(df.isnull().sum())
```

```python
print(df)
```

```python
print("Mean Age:",df["Age"].mean())
```

```python
print("Median Age:",df["Age"].median())
```

```python
df['Salary'] = df['Salary'].fillna(df['Salary'].median())
print(df)
```

```python
print("Mean Salary:",df["Salary"].mean())
```

```python
print("Median Salary:",df["Salary"].median())
```

```python
df['Salary'] = df['Salary'].fillna(df['Salary'].median())
print(df)
```
