```python
import numpy as np
import pandas as pd


data = {
    "Student":["A","B","C","D","E"],
    "Gender":["Male","Female","Male","Female","Male"],
    "Age":[19,np.nan,20,21,19],
    "StudyHours":[5,6,np.nan,4,7],
    "Attendance":[80,85,75,np.nan,90],
    "Marks":[75,82,68,55,np.nan],
    "Result":["Pass","Pass","Pass","Fail","Pass"]
}
df= pd.DataFrame(data)
print(df)

data = {
    'Name': ['Anu', 'Ravi', 'Priya', 'Kiran', 'Meena', 'Arjun'],
    'Age': [21, 22, None, 23, 21, None],
    'Salary': [25000, None, 30000, 28000, None, 32000],
    'Department': ['CSE', 'ECE', 'CSE', None, 'ISE', 'ECE'],
    'City': ['Mysore', 'Bangalore', 'Mysore', 'Hubli', 'Bangalore', 'Mysore']
}
df= pd.DataFrame(data)
print(df)
```

```python
print(df.isnull())
```

```python
print(df.isnull().sum())
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
print(df)
```

```python
df["StudyHours"] = df["StudyHours"].fillna(df["StudyHours"].mean())
print(df)
```

```python
df["Marks"] = df["Marks"].fillna(df["Marks"].median())
print(df)
```

```python
df["Attendance"]=df["Attendance"].ffill()
print(df)
```

```python
df["Result"] = df["Result"].map({
    "Pass" : 1,
    "Fail" :0
})

print(df)
```

```python
print(df.isnull().sum())
```

```python
print(df)
```

```python
print("Mean Age:",df["Age"].mean())
```

```python
print("Median Age:",df["Age"].median())
```
